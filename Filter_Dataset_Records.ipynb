```python
# Task 19: Filter Dataset Records

## Objective
Filter records from a dataset using one or more conditions
with Pandas.

## Deliverables
- At least three filtering examples
- Filtered DataFrames
- Explanation of each condition
- Number of records returned by each filter
```

```python
import pandas as pd
import numpy as np

data = {
    "student_id": [
        1001, 1002, 1003, 1004, 1005, 1006,
        1007, 1008, 1009, 1010, 1011, 1012
    ],
    "name": [
        "Aarav", "Diya", "Kabir", "Ananya",
        "Ishaan", "Meera", "Rohan", "Sara",
        "Vivaan", "Aditi", "Arjun", "Nisha"
    ],
    "department": [
        "Computer Science", "Electronics", "Mechanical",
        "Computer Science", "Civil", "Electronics",
        "Mechanical", "Computer Science", "Civil",
        "Electronics", "Mechanical", "Computer Science"
    ],
    "age": [20, 21, 20, 22, 21, 20, 22, 21, 20, 22, 21, 20],
    "marks": [86, 91, 74, 95, 68, 88, 79, 93, 72, 84, 77, 90],
    "attendance": [92, 88, 79, 96, 72, 90, 85, 94, 80, 86, 75, 91]
}

df = pd.DataFrame(data)

print("Student dataset created successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df)
```

```python
print("EXAMPLE 1: STUDENTS WITH MARKS >= 85")

filter_marks = df[df["marks"] >= 85]

display(filter_marks)

print("Number of matching records:", len(filter_marks))
```

```python
print("EXAMPLE 2: MARKS >= 80 AND ATTENDANCE >= 85")

filter_multiple = df[
    (df["marks"] >= 80) &
    (df["attendance"] >= 85)
]

display(filter_multiple)

print("Number of matching records:", len(filter_multiple))
```

```python
print("EXAMPLE 3: MARKS >= 90 OR ATTENDANCE >= 95")

filter_or = df[
    (df["marks"] >= 90) |
    (df["attendance"] >= 95)
]

display(filter_or)

print("Number of matching records:", len(filter_or))
```

```python
print("EXAMPLE 4: COMPUTER SCIENCE STUDENTS WITH MARKS >= 90")

filter_department = df[
    (df["department"] == "Computer Science") &
    (df["marks"] >= 90)
]

display(filter_department)

print("Number of matching records:", len(filter_department))
```

```python
summary = pd.DataFrame({
    "Filtering Example": [
        "Marks >= 85",
        "Marks >= 80 AND attendance >= 85",
        "Marks >= 90 OR attendance >= 95",
        "Computer Science AND marks >= 90"
    ],
    "Matching Records": [
        len(filter_marks),
        len(filter_multiple),
        len(filter_or),
        len(filter_department)
    ]
})

display(summary)
```

```python
df.to_csv("student_performance.csv", index=False)

print("student_performance.csv created successfully.")
```
